## Slide 16: Managing Context Weight

### 기술 핵심 내용 및 명세

#### Sweet Spot

#### Peripheral Noise and Outdated Logs

#### Latency

#### Relevance

#### System Logic and Architecture Docs

![Managing Context Weight 다이어그램](./images/m1_workflows_112.png)

> **발표자 노트:**
> - A common mistake in agentic workflows is 'Context Overload'—feeding the agent every single document in the repository.
> - We must manage 'Context Weight.' It’s a delicate balance: while adding more context generally improves task accuracy, it also increases processing latency.
> - Our goal is to tune the agent to prioritize high-signal dependencies while filtering out peripheral noise, ensuring we stay within that optimal performance sweet spot.
> - TRAINER TIP: Explain that "Context Weight" is the currency of agentic efficiency.
> - If the weight is too high, the agent spends all its time "reading" instead of "coding," which leads to deployment lag—the exact friction we are trying to solve.
