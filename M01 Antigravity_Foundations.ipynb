## 1. 개요 (Executive Overview)
본 문서는 엔터프라이즈 개발 환경에서 기존의 단순 로컬 AI 플러그인(Autocomplete 방식) 한계를 극복하고, Google Cloud의 에이전틱 플랫폼인 **Google Antigravity**를 도입하여 대규모 소프트웨어 엔지니어링 워크플로우를 자동화·자율화하는 핵심 기반 아키텍처를 다룸.

- **기존 한계(Legacy)**: IDE 내 라인 단위 자동완성, 로컬 리소스 종속, 프로젝트 간 컨텍스트 단절, 섀도우 코드(Shadow Code) 및 보안 감사 부재.
- **Antigravity 패러다임**: 거시적 태스크 위임(Delegating Outcomes), 클라우드 기반 서버사이드 하네스(Server-Side Harness), 소버린 보안 경계(Sovereign Security Boundaries), 결정론적 검증 아티팩트(Verifiable Artifacts & Execution Ledgers).
- **핵심 진입 접점(4대 인터페이스)**: Antigravity 2.0 (데스크톱 오케스트레이션 허브), `agy` CLI (터미널 네이티브), 프로그래밍 SDK (파이썬/TS 파이프라인 자동화), Agentic-Native Workspace (정밀 디버깅 작업공간).

| 핵심 주제영역 (Pillar) | 기존 AI 플러그인의 불편함 (Legacy Friction) | Antigravity 에이전틱 상태 (Target State) |
| :--- | :--- | :--- |
| **개발자 속도 (Velocity)** | 보일러플레이트(반복사용 기본 코드) 및 반복 작업으로 작업 시간 소모 | 자동 스캐폴딩(기본 구조 및 설정 생성) 및 `/goal` 매크로 태스크 위임 |
| **거버넌스 (Governance)** | 비공식 섀도우 코드 양산 및 수동 보안 감사 의존 | 프로젝트 ID 및 Cloud OAuth 기반 소버린 보안 경계 격리 |
| **신뢰성 (Trust)** | 블랙박스 AI의 코드 추측 및 검증 불가능한 변경 | 실행 원장(Execution Ledgers), Walkthrough, Diff 기반 검증 |
| **확장성 (Scale)** | 개발자 1:1 도구 병목 현상 | 중앙 집중식 Agent Manager를 통한 1:N 멀티 에이전트 오케스트레이션 |


### 운영 현황
#### 개발 현장의 실제 생산성 장벽

- **인지적 환상 (The Perception)**:
  - 단순 채팅창 도입만으로 무한한 생산성 향상이 가능할 것이라는 기대.
  - 별도 설정 없는 즉각적 코딩(Zero-config coding).
  - 프롬프트 입력 즉시 완성 상태('done' state) 도달.
- **현장의 기술적 현실 (The Reality)**:
  - **도구 파편화 및 인지 부하**: 콘솔, 에디터, 브라우저 간의 끝없는 컨텍스트 스위칭 발생.
  - **기본 작업의 과도한 오버헤드**: 배포 설정, 환경 구성, 보일러플레이트 작성에 시간 낭비.
  - **통합 프로젝트 컨텍스트 부재**: AI 도구가 저장소 전체 구조 및 다중 서비스 연관 관계를 이해하지 못함.

> **참고사항:**
> - 많은 조직이 AI 챗봇을 도입하면 개발 생산성이 즉시 10배 증가할 것이라 착각함.
> - 실제로는 도구 파편화, 컨텍스트 손실, 환경 격리 부재로 인해 개발자가 여전히 수동 배관 작업에 갇혀 있음. 가상의 Cymbal Bank 사례를 통해 이 문제의 실체와 해결방법을 규명함.


## 2. The Cymbal Bank Case Study
### 엔터프라이즈 실사례를 통한 개발 워크플로우 기반 정립

- **연구 대상**: Cymbal Bank (가상의 선도적 글로벌 금융 결제 처리 기관)
- **목표**: 실제 개발 워크플로우에 Antigravity 아키텍처를 그라운딩하여 실질적 문제 해결 프로세스 증명.

### 금융 결제 처리 기관의 직면 과제: 속도와 보안 규제의 충돌

- **산업군 (Industry)**: 엄격한 규제를 받는 금융 서비스 산업 (Highly regulated FSI).
- **기반 인프라 (Infrastructure)**: Google Cloud 표준화 인프라 환경.
- **핵심 미션 (Mission)**: 대규모 트랜잭션 이상 징후를 실시간 감지하는 자동화된 사기 탐지 API (Fraud-Detection API) 구축.
- **주요 난제 (The Challenge)**:
  - 결제 사기 탐지 지연으로 인한 트랜잭션 유출 및 손실 발생.
  - 비즈니스 요구는 신속한 API 출시를 요구하나, 금융 컴플라이언스 및 데이터 유출 방지 규정과의 상충 발생.
  - **핵심 딜레마**: 개발 속도(Velocity) vs 보안 규제 준수(Security Compliance).

> **참고사항:**
> - Cymbal Bank는 결제 승인 지연 및 이상 금융 거래 탐지 누수로 인해 즉각적인 API 고도화가 요구됨.
> - 인프라 엔지니어링 리드와 CIO는 혁신의 속도를 높이면서도 단 하나의 보안 규정도 위반할 수 없는 극단적인 제약에 직면함.

### 현재의 마찰 요인
#### 개발 전달을 지연시키는 3대 마찰 요인 및 비즈니스 결과

![현재의 마찰 요인](./images/m1_foundations_01.png)

1. **속도 차단 요인 (The Velocity Blockers)**:
   - 반복적인 보일러플레이트 코딩으로 개발자 가용 생산성 저하.
   - **비즈니스 영향**: 서비스 런칭 일정 지연 (Missed Launch Dates).
2. **보안 차단 요인 (The Security Blockers)**:
   - 검증되지 않은 외부 AI 도구 활용으로 인한 기업 자산 누출 및 거버넌스 위험.
   - **비즈니스 영향**: 금융 당국 규제 위반 과징금 부과 (Regulatory Fines).
3. **시스템 격차 (The System Gaps)**:
   - 팀 간 단절된 사일로(Silos)로 인해 엔드투엔드 오케스트레이션 불가.
   - **비즈니스 영향**: 통합 컨텍스트 부재로 인한 사기 탐지 정확도 저하 (Inaccurate Fraud Detection).

> **참고사항:**
> - Antigravity 아키텍처를 도입하기 전, 먼저 은행의 전달 체계를 마비시키는 병목 지점을 명확히 분리 분석해야 함. 속도, 보안, 시스템 고립이라는 3대 축으로 마찰 요인을 규명함.


### 속도 저해 요소
#### 개발 속도를 갉아먹는 세부 요인

- **보일러플레이트 부담 (Boilerplate Drag)**:
  - 암호화 핸들러, 인증 미들웨어, 라우팅 및 직렬화 등 반복적인 기본 코드 작성에 주당 근무 시간의 절반 소모.
- **도구 사일로 (Tooling Silos)**:
  - 클라우드 콘솔, 로컬 에디터, CI/CD 대시보드 간의 파편화된 워크플로우로 인해 빈번한 작업 전환 오버헤드 유발.
- **컨텍스트 윈도우 손실 (Context Window Losses)**:
  - 최신 API 명세, 스키마, 라이브러리 문서를 수동으로 검색하고 복사하는 과정에서 작업 집중도 분산.

> **참고사항:**
> - 팀 감사 결과, 인프라 리드를 포함한 핵심 개발자들이 업무 시간의 50%를 비즈니스 로직이 아닌 지루한 보일러플레이트 작업에 허비하고 있었음.


### 보안 저해 요소
#### 엔터프라이즈 보안 및 거버넌스 차단 요인

- **섀도우 코드 위험 (Shadow Code Risk)**:
  - 검증되지 않은 비공식 AI 도구를 사용하여 생성된 코드가 프로덕션 코드베이스에 유입될 위험 상존.
- **권한 과다 노출 (Permission Exposure)**:
  - 로컬 스크립트나 자동화 봇에 지나치게 넓은 권한(Over-privileged access)이 부여되어 침해 사고 위험 가중.
- **거버넌스 공백 (Governance Gaps)**:
  - AI가 생성한 변경 사항에 대해 누가, 왜, 어떤 프롬프트와 컨텍스트로 코드를 수정했는지 추적 가능한 감사 추적(Audit Trails) 부재.

> **참고사항:**
> - CIO의 최대 우려는 데이터 유출과 섀도우 코드임. 소스코드와 민감한 은행 스키마가 외부 LLM 학습에 활용되거나 감사 기록 없이 배포되는 것은 규제 기관의 철퇴를 부르는 치명적 요인임.


### 시스템 격차
#### 단절된 아키텍처로 인한 시스템적 격차

- **사일로화된 실행 (Siloed Execution)**:
  - 개별 개발자의 로컬 에디터에 설치된 고립된 플러그인 형태로 동작하여 팀 전체 워크플로우와 연결되지 않음.
- **공유 메모리 부재 (No Shared Memory)**:
  - 인프라 변경 내역, 데이터 모델 수정 사항, 보안 정책 등이 팀원 간에 공유되지 않고 휘발됨 (Zero cross-team awareness).
- **중앙 오케스트레이션 결여 (Orchestration Absence)**:
  - 멀티 에이전트를 모니터링하고 정책을 강제하며 태스크 진행 상황을 일괄 조율하는 컨트롤 타워 부재.

> **참고사항:**
> - 개별 개발자가 각자의 데스크톱에서 독립적인 AI 도구를 쓰면 조직 전체 차원의 학습과 시너지는 전혀 발생하지 않음. 중앙 오케스트레이션과 공유 메모리가 반드시 필요함.


## 3. 에이전틱 플랫폼 패러다임
### 에이전틱 플랫폼 패러다임과 생태계 아키텍처 정의

- **핵심 목표**: 단순 도구 모음을 넘어선 단일 통합 엔진 아키텍처 분석.
- **패러다임 전환**: 미시적 라인 코딩에서 거시적 시스템 결과 위임으로의 전환 원리 고찰.

> **참고사항:**
> - Cymbal Bank의 마찰 지점을 확인했으니 이제 아키텍처적 해법으로 전환함. 에이전틱 플랫폼 패러다임이 어떻게 이러한 마찰을 근본적으로 제거하는지 살펴봄.


### Antigravity 정의
#### Google Cloud의 통합 에이전트 우선(Agent-First) 개발 플랫폼

- **플랫폼 정의**:
  - Google Cloud의 통합 에이전트 우선 개발 플랫폼 (Unified, Agent-First Development Platform).
  - 여러 도구의 파편화된 결합이 아닌, 단일 백엔드 엔진 기반의 통합 아키텍처.
- **핵심 전환 목표**:
  - 세부적인 한 줄 코드 구현(Granular line implementation)에서 거시적 태스크 위임(Macroscopic task delegation)으로 전환.
- **단일 플랫폼 통합 체계**:
  - **UI / Desktop App** (Antigravity 2.0)
  - **CLI** (`agy` 터미널 래퍼)
  - **Programmatic SDK** (파이썬 / TypeScript 바인딩)
  - **Managed Agents API** (중앙 집중식 에이전트 수명주기 관리)
- **통합 컨텍스트 (Unified Context)**:
  - 모든 인터페이스가 단일 공유 컨텍스트를 참조하여 도구 간 전환 시 발생하는 스코프 손실 제거.

> **참고사항:**
> - Antigravity는 별개의 도구 컬렉션이 아님. 모든 개발자 접점이 동일한 중앙 엔진과 컨텍스트 저장소를 공유하므로, 터미널에서 시작한 작업을 데스크톱 UI나 CI/CD 파이프라인에서 그대로 이어받아 제어할 수 있음.


### 플러그인 모델에서 벗어나기
#### 레거시 IDE 플러그인 모델 탈피: 3대 차원 비교 분석


| 비교 차원 (Dimension) | 기존 레거시 코드 플러그인 (Legacy Plugins) | Antigravity 엔진 (Antigravity Engine) |
| :--- | :--- | :--- |
| **운영 모델<br>(Operating Model)** | 에디터 자동완성 방식. 로컬 파일 컨텍스트 내에서 라인 단위의 수동적(Passive) 제안 수행. | 활성 서버사이드 에이전트 하네스. 복잡한 개발 작업을 자율적으로 계획 및 실행. |
| **실행 범위<br>(Execution Scope)** | 타이핑 패턴에 기반한 글자/단어 단위 추측(Character-by-character guessing). | 엔드투엔드 소프트웨어 엔지니어링 워크플로우를 조율하는 거시적 계획(Macro Task Planning). |
| **실행 환경<br>(Environment)** | 로컬 IDE 프로세스 내에서 직접 실행되며, 로컬 시스템 메모리 및 CPU 리소스 한계에 종속. | 고성능 클라우드 인프라 상의 격리된 일회성(Ephemeral) Linux 샌드박스에서 안전하게 실행. |

> **참고사항:**
> - Antigravity를 이해하려면 기존 자동완성 플러그인 방식의 고정관념을 버려야 함. 기존 방식이 타이핑하는 손끝을 쳐다보며 다음 단어를 맞추는 것이라면, Antigravity는 클라우드 샌드박스에서 빌드, 테스트, 리팩토링 전 과정을 자율 수행하는 독립 엔지니어링 에이전트임.


### 사고의 전환
#### 개발자의 핵심 멘탈 모델 전환: 위임과 검증

![사고의 전환](./images/m1_foundations_02.png)

- **01. 결과 중심 정의 (Focus)**: 어떻게(How) 구현할지가 아니라 무엇(What)을 달성할지와 어떤 제약(Constraints)을 지킬 것인지 선언.
- **02. 거시적 위임 (Delegation)**: 단순 문법 완성이 아닌, 특정 마이크로서비스 모듈 스캐폴딩 및 암호화 연동 전체를 에이전트에 위임.
- **03. 구조적 검증 (Verification)**: 블랙박스 실행 결과를 무비판적으로 수용하지 않고 실행 원장(Ledger), 변경 내역(Diff), 테스트 결과를 통해 검증.

> **참고사항:**
> - 이제 개발자는 키보드로 코드를 한 줄씩 타이핑하는 단순 작업자에서, 시스템 목표를 정의하고 결과 아티팩트를 승인하는 '소프트웨어 아키텍트'이자 '디렉터'로 역할이 전환됨.


### 서버사이드 하네스
#### 고성능 서버사이드 하네스 아키텍처

- **핵심 엔진 (Core Engine)**:
  - Google Gemini 3.5 Flash 플랫폼 로직 및 최신 추론 엔진과 공동 최적화(Co-optimized)되어 초고속 응답 보장.
- **로컬 머신 부하 제로 (Local Footprint)**:
  - 로컬 환경에서는 초경량 프로세스만 실행되며, 무거운 컴파일, 의존성 다운로드, 테스트 실행 등은 전부 클라우드로 오프로드됨.
- **격리된 클라우드 연산 (Compute)**:
  - 태스크마다 격리된 일회성(Ephemeral) 서버사이드 환경이 프로비저닝되어 로컬 OS 오염 방지 및 무결성 유지.

![서버사이드 하네스 아키텍처](./images/m1_foundations_03.png)

> **참고사항:**
> - 로컬 노트북의 팬이 돌아가거나 배터리가 닳지 않음. 대규모 빌드와 테스트 러너가 모두 Google Cloud의 격리된 서버 환경에서 수 초 내에 고속 병렬 처리됨.


### 소버린 보안 경계
#### 기업 자산과 IP를 완벽히 격리하는 소버린 보안 경계

- **엔터프라이즈 신원 연계 (Identity)**:
  - 모든 에이전트 세션은 기업용 Google Cloud OAuth 및 IAM과 직접 바인딩되어 사용자 권한 범위 내에서만 인가됨.
- **프로젝트 단위 완전 격리 (Isolation)**:
  - 지정된 Google Cloud `Project ID` 경계 내부에서만 태스크가 실행되며 타 프로젝트나 외부로의 데이터 유출 차단.
- **지적재산권 보호 및 비학습 보장 (Governance)**:
  - 고객의 소스코드, 스키마, 페이로드는 완전히 차단(Walled off)되며 일반 공개 모델 학습(Public training)에 일체 사용되지 않음.
- **컨텍스트 페이로드 및 검증된 추론 (Validated Inference)**:
  - 안전하게 검증된 컨텍스트 페이로드만을 전달하여 악의적 프롬프트 인젝션 및 무단 호출 차단.

> **참고사항:**
> - Cymbal Bank CIO의 최대 우려를 원천 해결하는 아키텍처임. 인가된 프로젝트 ID 내부에서만 코드가 오가며, 엔터프라이즈 계약에 따라 기업의 코드는 모델 학습에 절대 사용되지 않음.


### 검증 가능한 에이전틱 산출물
#### 블랙박스를 해체하는 결정론적 검증 아티팩트 및 4단계 루프

![검증 가능한 에이전틱 산출물](./images/m1_foundations_04.png) 

1. **목표 정의 (Define)**: 개발자가 목표와 아키텍처 제약 조건을 명시함.
2. **실행 계획 수립 (Plan)**: 에이전트가 단계별 실행 경로와 영향도를 담은 구조적 계획(Implementation Plan)을 생성함.
3. **인간 중심 검증 (Review)**: 계획서, 실행 로그, 파일별 Diff가 포함된 **실행 원장(Execution Ledger)** 을 사람이 직접 검토 및 승인함.
4. **저장소 커밋 (Commit)**: 인간의 명시적 승인이 완료된 검증 코드만 리포지토리에 최종 통합 및 커밋됨.

> **참고사항:**
> - 에이전트가 코드를 백그라운드에서 임의로 수정하여 덮어쓰지 않음. 모든 작업은 Walkthrough와 Diffs가 담긴 실행 원장(Ledger) 아티팩트로 출력되며, 개발자의 최종 승인을 거쳐야만 프로덕션에 병합됨.


## 4. Antigravity 생태계 4대 핵심 인터페이스
### 4대 진입 접점 

| 번호 | 인터페이스 명칭 | 주요 성격 및 타겟 워크플로우 |
| :--- | :--- | :--- |
| **01** | **Antigravity 2.0 (Desktop)** | **오케스트레이션 허브 & 중앙 집중식 Agent Manager UI**<br>거시적 다중 에이전트 계획 수립, 정책 관리 및 시각적 조율 |
| **02** | **Antigravity CLI (`agy`)** | **터미널 네이티브 셸 래퍼 (Go 언어 기반)**<br>기존 개발자 셸 환경에서 고속 실행 및 리포지토리 자동 감지 |
| **03** | **Programmatic SDK** | **파이썬 / TypeScript 프로그래밍 라이브러리**<br>CI/CD, GitHub Actions, Jira 등 자동화 백엔드 파이프라인 내 에이전트 결합 |
| **04** | **Agentic-Native Workspace** | **에이전트 특화 정밀 개발 환경**<br>다중 리포지토리(Multi-repo) 정밀 디버깅, 즉각적인 코드 수정 피드백 루프 제공 |

> **참고사항:**
> - 개발자마다 작업 스타일이 다름. 인프라 리드는 데스크톱 허브를 선호하고, 코어 개발자는 CLI를, 플랫폼 엔지니어는 SDK를 사용함. 동일한 엔진이 4가지 인터페이스 뒤에서 일관되게 동작함.


### Surface 1 - Antigravity 2.0 (Desktop App)
#### 거시적 오케스트레이션 및 에이전트 관리 컨트롤 타워

- **독립형 데스크톱 앱 (Standalone Desktop App)**:
  - 시각적 실행 파이프라인과 로컬 제어 기능을 결합하여 전체 작업 진행 상황에 대한 완전한 가시성 제공.
- **중앙 집중식 에이전트 매니저 (Centralized Agent Manager)**:
  - 실행 중인 다중 에이전트 목록 실시간 모니터링.
  - 전사 보안 정책 및 가드레일 강제 적용.
  - 에이전트가 생성한 실행 계획 및 아티팩트 일괄 관리.
- **강력한 디렉티브 프로토콜 (Directive Protocols)**:
  - `/goal`: 복잡한 매크로 계획 수립 및 장기 실행 미션 개시.
  - `/grill-me`: 에이전트가 역질문 인터뷰를 수행하여 모호한 요구사항과 아키텍처 결정을 사전에 명확화.

> **참고사항:**
> - 인프라 리드가 주로 사용하는 커맨드 센터임. 여러 에이전트가 동시에 실행될 때 각 에이전트의 상태와 산출물을 시각적으로 한눈에 파악하고 조율할 수 있음.


### Surface 2 - The Antigravity CLI (agy)
#### 터미널 중심 고성능 셸 래퍼 및 대역폭 최적화

- **터미널 네이티브 통합 (Seamless Workflow)**:
  - 별도 GUI 실행 없이 기존 Bash, Zsh, PowerShell 등 개발자 터미널에 완벽히 임베딩.
- **저장소 인텔리전스 (Contextual Awareness)**:
  - 실행 디렉토리의 프로젝트 구조, 종속성 파일(`go.mod`, `package.json`, `pom.xml`), 로컬 Git 브랜치 상태 자동 감지.
- **70% 토큰 절감 필터 (Bandwidth Optimization)**:
  - 방대한 빌드 출력물, 패키지 다운로드 로그, 불필요한 스택 트레이스 등의 '터미널 노이즈'를 소스 레벨에서 지능적으로 필터링하여 컨텍스트 윈도우 보존 및 응답 속도 극대화.

> **참고사항:**
> - 터미널 중심의 엔지니어에게 최적화된 도구임. 특히 70% 토큰 절약 필터는 방대한 빌드 로그로 인해 에이전트가 이전 컨텍스트를 잊어버리는 현상을 원천 방지함.


### Surface 3 - The Programmatic SDK
#### 파이프라인 자동화 및 이벤트 기반 헤드리스 에이전트 통합

- **지원 언어**: Native Python 및 TypeScript 지원으로 기존 소프트웨어 아키텍처에 자연스럽게 융합.
- **자율형 백엔드 워크플로우**: 이벤트 기반 인프라 자동화 및 헤드리스(Headless) 에이전트 루프 구축.
- **엔터프라이즈 에코시스템 연동**: GitHub, Jira, ArgoCD, Cloud Build 등 CI/CD 파이프라인과 직접 연계.

##### 예제 코드: GitHub 이벤트 연동 자동 테스트 에이전트 배포
```python
import antigravity
from typing import Dict, Any

# Google Cloud 자율 코어 엔진에 연결
sdk = antigravity.connect_cloud()

def handle_github_event(payload: Dict[str, Any]) -> bool:
    """저장소 푸시 이벤트를 처리하여 백엔드 자동화 테스트 에이전트 파이프라인을 트리거함."""
    event_type = payload.get("X-GitHub-Event")
    repo_name = payload.get("repository", {}).get("full_name")
    print(f"Event: {event_type} on repository: {repo_name}")
    
    # 자율 테스트 에이전트 파이프라인 배포 및 실행
    return sdk.deploy_backend_workflow(repo=repo_name, mode=True)
```

> **참고사항:**
> - 데이터/백엔드 리드는 SDK를 활용하여 PR 생성 시 자동으로 단위 테스트를 작성하고 보안 취약점을 검사하는 에이전트 워크플로우를 구축함.


### Surface 4 - Agentic-Native Workspace
#### 멀티 리포지토리 정밀 제어 및 에이전트 네이티브 작업공간

- **정밀 에이전틱 컨텍스트 (Precision Focus)**:
  - 에이전트와 코드 간의 즉각적인 피드백 루프를 지원하여 심층 시각적 디버깅 및 세밀한 코드 리팩토링 지원.
- **다중 리포지토리 통합 아키텍처 (Multi-Repo Architecture)**:
  - 마이크로서비스 환경에서 여러 저장소에 걸친 구조적 변경 사항을 단일 통합 작업창에서 조율.
- **기존 개발자 경험 보존 (Developer Mechanics)**:
  - 네이티브 IDE의 단축키와 터미널 머슬 메모리를 그대로 계승하여 별도 학습 비용 없이 자연스럽게 적응.

> **참고사항:**
> - 서비스가 여러 리포지토리로 분할되어 있을 때 한쪽의 API 스키마 변경이 다른 리포지토리에 미치는 영향을 단일 뷰에서 추적하고 디버깅할 수 있는 에이전트 특화 작업공간임.


### Cymbal Mission에 매핑
#### 엔지니어링 의도와 규모에 따른 인터페이스 매핑 매트릭스

![인터페이스 매핑](./images/m1_foundations_05.png)

- **핵심 원칙**: 엔지니어링 태스크의 **의도(Intent)** 와 **규모(Scale)** 에 정확히 일치하는 인터페이스를 선택하여 실행.

> **참고사항:**
> - 모든 상황에 한 가지 도구만 고집할 필요 없음. 아키텍처 스캐폴딩은 데스크톱 허브에서, 개별 개발은 CLI에서, 파이프라인 자동화는 SDK에서 수행하는 체계적 분업이 가능함.


## 5. 마이그레이션 워크스루
### 실전 마이그레이션 워크스루: Cymbal Bank DEMO

- **주요 내용**: 단편적 레거시 워크플로우에서 Antigravity 패러다임으로의 단계적 전환 절차.
- **실전 데모**: 태스크 개시부터 거버넌스 검증, 인터뷰 프로토콜, 아티팩트 승인까지의 전 과정 시뮬레이션.

> **참고사항:**
> - 도구와 보안 모델을 파악했으니 이제 실제로 마이그레이션하는 3단계 프레임워크와 실시간 시뮬레이션 워크플로우를 진행함.


### 3단계 도입 프레임워크
#### 체계적인 3단계 도입 프레임워크 (Baseline → Mapping → Operationalization)

![3단계 도입 프레임워크](./images/m1_foundations_06.png)

- **원칙**: 단순한 '리프트 앤 시프트(Lift & Shift)' 방식이 아닌, 거버넌스 확립에서 출발하여 운영 자율화로 나아가는 점진적 프레임워크 준수.

> **참고사항:**
> - 도입 과정은 보안 경계를 세우고(Baseline), 병목 업무를 매핑하며(Mapping), 최종적으로 SDK로 자동화하는(Operationalization) 3단계로 엄격히 관리되어야 리스크가 없음.


### Step 1 - Baseline
#### 소버린 보안 경계 수립 및 거버넌스 인가

- **실행 주체**: CIO 및 인프라 보안팀.
- **핵심 조치**:
  - 엔터프라이즈 Google Cloud OAuth 세션 승인 및 서비스 계정 연동.
  - 사기 탐지 API 프로젝트 ID(`project-cymbal-fraud`)에 대한 엄격한 리소스 경계 지정.
  - 외부 모델 학습 금지 및 네트워크 아웃바운드 차단 정책 확정.
- **효과**: 코드가 한 줄도 작성되기 전에 완전한 보안 및 컴플라이언스 인프라 확보.

> **참고사항:**
> - 첫걸음은 언제나 보안임. CIO가 승인한 보안 경계가 확립되어야만 개발팀이 안심하고 에이전트를 가동할 수 있음.


### Step 2 - Mapping
#### 고마찰 보일러플레이트 식별 및 인터페이스 할당

- **실행 주체**: 인프라 리드 및 코어 개발팀.
- **핵심 조치**:
  - 암호화 훅(Crypto hooks), 엔드포인트 라우팅, 스키마 유효성 검증 등 개발 속도를 저해하는 고마찰 보일러플레이트 과제 선정.
  - 각 과제를 처리할 최적 인터페이스(`agy` CLI 및 Antigravity 2.0 데스크톱) 선정.
- **효과**: 개발팀이 반복적 기본 세팅 작업에서 해방되어 핵심 비즈니스 로직에 집중 가능.

> **참고사항:**
> - 보안 경계가 잠기면, 이제 개발자의 손발을 묶고 있던 지루한 보일러플레이트 태스크를 선별하여 에이전트에 위임할 준비를 마침.


### Step 3 - Operationalization
#### SDK 기반 운영 자동화 및 이벤트 기반 에이전트 확장

- **실행 주체**: 데이터/백엔드 리드 및 플랫폼 엔지니어링팀.
- **핵심 조치**:
  - 수동으로 에이전트를 트리거하던 방식을 벗어나, Programmatic SDK를 활용하여 CI/CD 파이프라인 및 Git Webhook에 결합.
  - 코드 푸시 시 자동 테스트 작성, 취약점 스캔, 문서화가 자율적으로 실행되는 루프 완성.
- **효과**: 조직 전체 개발 라이프사이클의 유기적 확장 및 지속적 품질 향상 달성.

> **참고사항:**
> - 마지막 단계는 운영 자율화임. 수동 호출을 완전히 배제하고 이벤트가 발생할 때마다 에이전트가 알아서 검증과 배포를 돕는 자율 루프를 완성함.

> **참고사항:**
> - 이론을 실제 워크플로우로 시연함. 교육 환경에 맞추어 태스크 시작부터 최종 승인까지의 전 과정을 단계별로 시뮬레이션함.


### 4단계 프로세스
#### 에이전틱 실행의 4단계 프로세스

![4단계 프로세스](./images/m1_foundations_07.png)

- **Part 1 (The Setup)**:
  - 1. 정책 검증 (Policy Check)
  - 2. 보안 격리 (Secure Isolation)
- **Part 2 (The Execution)**:
  - 3. 의도 발견 (Intent Discovery)
  - 4. 아티팩트 생성 (Artifact Generation)

> **참고사항:**
> - 시뮬레이션은 4개의 논리적 단계로 전개됨. 준비 단계(Setup)에서 거버넌스를 강제하고, 실행 단계(Execution)에서 요구사항 인터뷰와 아티팩트 검증을 완료함.


### Part 1 - The Setup | Task Initiation & Governance
#### 태스크 개시 및 거버넌스 검증 루프

##### 터미널 실행 화면 및 거버넌스 로그
```bash
$ agy /goal "암호화 훅을 적용한 이상 거래 탐지 API의 기본 구조를 구축하세요."
⠋ Analyzing project project-cymbal-fraud...
✔ Identity Verified (OAuth)
✔ Project Boundary Locked
⠋ Spawning Agent: scaffold-agent-01
```

- **01 정책 검증 (Policy Check)**
  - 기업용 OAuth 신원과 대상 `Project ID`의 권한을 즉시 상호 검증.
- **02 보안 격리 (Secure Isolation)**
  - 에이전트의 모든 파일 조작과 연산을 사전에 허가된 소버린 보안 경계 내부로 제한하며, 무거운 연산은 클라우드 서버사이드 하네스로 오프로드.
- **거버넌스 차단 루프 (The Governance Loop)**
  - CIO가 정의한 소버린 보안 경계 내에 에이전트가 완벽히 안착되었음이 확인될 때까지 코드 생성 실행이 일시 차단됨.

> **참고사항:**
> - `/goal` 명령을 입력했을 때 시스템은 곧바로 코드를 생성하지 않음. 신원과 프로젝트 경계를 먼저 검증하는 거버넌스 루프가 최우선으로 작동함.


### Part 2 - The Execution | Agentic Action & Artifacts
#### 에이전틱 실행 및 검증 아티팩트 생성

- **03 의도 발견 (Intent Discovery)**:
  - 에이전트가 `/grill-me` 인터뷰 프로토콜을 자율적으로 발동하여 Tom에게 암호화 알고리즘 요구사항(AES-256-GCM vs RSA)을 역질문하고 스펙을 확정함.
- **04 아티팩트 생성 (Artifact Generation)**:
  - 코드를 임의 커밋하지 않고 실행 원장(Execution Ledger)을 컴파일하여 출력.
  - 인간 엔지니어가 구현 계획(Implementation Plan)과 코드 변경 내역(Diffs)을 사전에 면밀히 검토 및 승인.

##### 에이전트 매니저(Agent Manager) 검토 목록
- **Active Agents**: `scaffold-agent-01`
- **검토 대상 파일 Diff**:
  - `Update crypto_lib.go` (암호화 라이브러리 갱신)
  - `Patch auth_service.ts` (인증 미들웨어 패치)
  - `Generate test_suite.py` (단위/통합 테스트 스위트 생성)

> **참고사항:**
> - `/grill-me`를 통해 요구사항의 모호함을 사전에 제거하고, 생성된 코드는 Diff 형태로 검토 창에 표시됨. Tom이 승인하기 전까지는 프로덕션에 단 한 줄도 반영되지 않아 감사 요구사항을 완벽히 충족함.


## 6. 요약 - Antigravity 성공 로드맵
### Cymbal Bank 도입 성과 맵: 4대 주제영역 전후 비교

| 운영 주제영역 (Operational Pillar) | 레거시 마찰 상태 (Legacy Friction State) | 에이전틱 도입 상태 (Antigravity Target State) |
| :--- | :--- | :--- |
| **개발자 속도<br>(Developer Velocity)** | 수동 보일러플레이트 작성으로 업무 시간 50% 낭비 | 자율 스캐폴딩 및 `/goal` 매크로 태스크 위임으로 출시 주기 단축 |
| **보안 거버넌스<br>(Governance)** | 섀도우 코드 위험 및 수동 감사 의존에 따른 규제 위반 노출 | 프로젝트 ID 및 Cloud OAuth 기반 소버린 보안 경계 완전 격리 |
| **신뢰성<br>(Trust)** | 블랙박스 AI의 추측성 코드 생성 및 검증 부재 | 실행 원장(Execution Ledgers) 및 구조적 Diff 검증 체계 확립 |
| **확장성<br>(Scale)** | 1:1 도구 인터랙션 병목 현상 | 중앙 집중식 Agent Manager 기반 1:N 멀티 에이전트 조율 체계 완성 |

> **참고사항:**
> - Cymbal Bank는 단순한 도구를 구매한 것이 아니라 운영 체계를 바꿈. '희망 기반 보안'을 '소버린 경계'로, '블랙박스 추측'을 '검증 가능한 원장'으로 대체함.


### 핵심 요약
#### 모듈 1 핵심 요약: 4대 핵심 원칙

1. **거버넌스 기반 확장 (Governed Scale)**:
   - 속도를 위해 보안을 희생하지 않으며, 신원·컨텍스트·IP를 보호하는 소버린 경계(Sovereign Boundaries)를 엄격히 적용함.
2. **에이전트 패러다임 (Agent Paradigm)**:
   - 라인 단위 타이핑 추측에서 벗어나 통합 서버사이드 하네스를 통해 결과(Outcomes) 자체를 위임함.
3. **통합 툴킷 (Unified Toolkit)**:
   - 엔지니어링 의도에 맞는 최적 인터페이스(Desktop, CLI, SDK, Workspace)를 유연하게 선택함.
4. **결정론적 검증성 (Verifiability)**:
   - 블랙박스 AI 리스크를 배제하고 실행 원장(Execution Ledgers)과 감사 가능한 아티팩트로 완벽한 신뢰를 구축함.

> **참고사항:**
> - 반드시 기억해야 할 4가지 축 - 거버넌스 기반 확장, 결과 위임 패러다임, 통합 인터페이스, 검증 가능한 아티팩트가 엔터프라이즈 성공의 열쇠임.
