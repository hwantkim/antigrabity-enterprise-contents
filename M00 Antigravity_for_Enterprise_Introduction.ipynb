## 1. 개요 및 목적
- 본 문서는 Google Cloud의 엔터프라이즈 에이전트 개발 플랫폼인 **Antigravity**의 아키텍처 개요, 첫 온보딩 가이드, 핵심 작동 메커니즘을 상세히 정리한 기술 문서임.
- 기존 단순 코드 자동완성 수준의 도구를 넘어, 엔터프라이즈 환경에서 자율적 소프트웨어 엔지니어링을 수행하는 **통합 서버사이드 에이전틱 하네스(Server-side Agentic Harness)**로의 패러다임 전환을 설명함.
- CLI(`agy`), IDE 확장, 웹 기반 Command Center 전반에 걸친 아키텍처와 **Prompt-to-Diff** 3단계 라이프사이클, 토큰 이코노믹스 최적화 방안을 포괄함.

## 2. 주요 아키텍처 핵심 요약
| 핵심 기둥 | 기술 메커니즘 | 엔터프라이즈 가치 |
| :--- | :--- | :--- |
| **Agentic Harness** | 자연어 요구사항을 멀티 파일 엔지니어링 액션으로 직접 변환 | 작업 단위의 end-to-end 개발 자동화 |
| **Deterministic Control** | 추측성 변경 대신 구조화된 실행 계획(Plan), 원장, 비주얼 diff 제공 | 변경 내역 사전 검증 및 거버넌스 확보 |
| **Server-Side Runtimes** | 무거운 빌드, 분석, 테스트 연산을 클라우드 샌드박스로 오프로드 | 로컬 자원 절약 및 일관된 실행 환경 보장 |
| **Token Optimization** | 타겟 컨텍스트 참조 및 동적 파일 디스커버리 커넥터 활용 | 128k 윈도우 중 ~2.6%만 사용하여 지연 최소화 및 정확도 극대화 |

  - 교육 진행 전 반드시 원본 슬라이드의 사본을 생성하여 활용함.

  - 참석자 파일 공유 시 다운로드, 인쇄, 복사 제한 옵션을 적용함.

## Slide 09: Antigravity 개요 - 패러다임 전환과 엔지니어링 비전

### 1. 패러다임 전환 (The Shift)
- **기존 방식**: 로컬 개발 환경에서 단일 라인 수준의 코드를 단순 추천하는 고립된 코드 어시스턴트(Isolated Code Assistants)
- **Antigravity 방식**: 엔터프라이즈 에이전트 플랫폼에 의해 구동되는 **통합 서버사이드 에이전틱 하네스(Unified, Server-Side Agentic Harness)**로 전환

### 2. 플랫폼 비전 (The Vision)
- 미가공 생성형 파운데이션 모델(Raw Generative Models)과 실제 **자율 소프트웨어 엔지니어링(Autonomous Software Engineering)** 사이의 간극을 연결함.
- 단순 텍스트 완성을 넘어, 다중 파일에 걸친 변경 사항을 안전하고 결정론적이며 투명하게 오케스트레이션함.

```mermaid
graph LR
    A["Isolated Code Assistants<br/>(단일 라인 코드 자동완성)"] -->|Paradigm Shift| B["Unified Agentic Harness<br/>(서버사이드 에이전트 플랫폼)"]
    B --> C["Autonomous Software Engineering<br/>(멀티 파일 작업 & 거버넌스 보장)"]
```

> **발표자 노트:**
> - 우리가 나아갈 큰 그림을 이해하는 것이 매우 중요함.
> - 단순히 한두 줄의 코드를 채워주는 어시스턴트에서 벗어나, 대규모 소프트웨어 변경을 안전하게 처리하는 통합 에이전트 하네스로 진화하는 핵심 전환점을 강조함.


## Slide 10: Antigravity의 3대 핵심 아키텍처 기둥

### 플랫폼 핵심 구성 요소
1. **Agentic Harness (에이전틱 하네스)**:
   - 자연어 요구사항(Natural Language Intent)을 파싱하여 멀티 파일 엔지니어링 작업 단위로 즉각 변환함.
2. **Deterministic Control (결정론적 통제 체계)**:
   - 추측이나 블랙박스형 생성 방식을 배제하고, 구조화된 계획(Reasoning Plan), 실행 원장(Execution Ledger), 시각적 diff(Visual Diff)를 제공함.
3. **Server-Side Runtimes (서버사이드 런타임)**:
   - 리소스를 대량 소모하는 코드 분석, 빌드, 컴파일, 테스트 작업을 로컬 워크스테이션에서 클라우드로 완전 오프로드함.

| 아키텍처 컴포넌트 | 담당 역할 | 기술 구현 및 이점 |
| :--- | :--- | :--- |
| `Agentic Harness` | 의도 해석 및 멀티 파일 태스크화 | 컨텍스트 기반 의존성 추적 및 변경 파일 자동 식별 |
| `Deterministic Control` | 실행 안정성 및 승인 게이트 | 사전 계획 수립 후 인간 검토 승인(Human-in-the-loop) 강제 |
| `Server-Side Runtimes` | 연산 오프로딩 | 개발자 PC 사양 제약 극복 및 격리된 클라우드 샌드박스 보장 |

![What is Antigravity Architecture](./images/antigravity_intro_12.png)

> **발표자 노트:**
> - Antigravity가 내부적으로 어떻게 작동하는지 3가지 핵심 기둥을 설명함.
> - 첫째는 자연어를 다중 파일 엔지니어링 작업으로 바꾸는 에이전틱 하네스,
> - 둘째는 무거운 컴파일/분석 작업을 클라우드에서 처리하는 서버사이드 런타임,
> - 셋째는 시행착오 대신 명확한 계획과 diff로 개발자가 완전한 제어권을 쥐게 하는 결정론적 제어 메커니즘임.


## Slide 11: Zero-to-Prompt 3단계 초고속 온보딩

### 온보딩 워크플로우 명세
- **단계 1: 설치 (Installation)**
  - 경량 로컬 CLI 도구인 `agy` 설치 또는 선호하는 IDE 확장(Extension)을 설정함.
- **단계 2: 기업 SSO 연동 (Enterprise SSO)**
  - 조직의 표준 IDP(Identity Provider) 및 IAM 인증 체계를 통해 안전하게 엔터프라이즈 계정 인증을 완료함 (`agy login`).
- **단계 3: 프로젝트 바인딩 (Project Linking)**
  - 현재 로컬 작업 디렉토리를 승인된 Google Cloud 프로젝트 ID에 바인딩하여 클라우드 리소스 및 에이전트 권한을 연계함.

```mermaid
flowchart LR
    Step1["01. Installation<br/>(CLI agy / IDE 확장)"] --> Step2["02. Enterprise SSO<br/>(조직 IDP 보안 인증)"]
    Step2 --> Step3["03. Project Linking<br/>(Google Cloud 프로젝트 ID 바인딩)"]
    Step3 --> Ready["Ready to Prompt!<br/>(즉시 에이전틱 프롬프트 실행 가능)"]
```

> **발표자 노트:**
> - 초기 설정에 5분도 걸리지 않음을 안내함.
> - CLI나 IDE 확장을 설치하고 기업 단일 로그인(SSO)을 마친 뒤, 로컬 프로젝트 디렉토리를 클라우드 프로젝트 ID에 바인딩하면 즉시 첫 에이전틱 프롬프트를 실행할 준비가 완료됨.


## Slide 12: 에이전틱 워크스페이스의 내부 구조 (Anatomy of an Agentic Workspace)

### 3대 핵심 사용자 인터페이스 요소
1. **01 // Intent Input Panel (의도 입력 패널)**:
   - 자연어 프롬프트를 통해 아키텍처 목표를 정의하는 핵심 인터페이스임.
   - 특정 파일/컴포넌트 멘션을 위한 `@` 문법, 사전 정의 액션 실행을 위한 `/` 슬래시 커맨드를 지원함.
   - 상단에서 기반 LLM 모델(예: `Gemini 3.6 Flash (High)`)을 선택할 수 있음.
2. **02 // Context Drawer (컨텍스트 서랍)**:
   - 에이전트가 현재 참조하고 있는 파일, 문서, 티켓 목록을 가시적으로 노출함.
   - 대화 내역(`Conversation History`), 예약된 작업(`Scheduled Tasks`), 프로젝트(`Projects`) 탭을 통한 컨텍스트 전환을 지원함.
3. **03 // Open IDE Launcher (IDE 즉시 실행 버튼)**:
   - 우측 상단 퀵 액세스 컨트롤을 통해 현재 작업 중인 워크스페이스를 기본 로컬 IDE(VS Code 등)에서 즉시 실행하고 심층 편집을 이어갈 수 있음.

| UI 컴포넌트 | 화면 위치 | 핵심 기능 |
| :--- | :--- | :--- |
| `Intent Input Panel` | 하단/중앙 | 요구사항 자연어 입력, `@` 컨텍스트 멘션, `/` 명령어, 모델 토글 |
| `Context Drawer` | 좌측 패널 | 활성 파일 추적, 세션 히스토리, 백그라운드 태스크 관리 |
| `Open IDE Launcher` | 우측 상단 | 로컬 개발 환경(IDE) 즉시 연동 및 딥 다이브 코드 수정 |

> **발표자 노트:**
> - 워크스페이스를 열었을 때 마주하게 되는 3대 핵심 UI 영역을 설명함.
> - 프롬프트를 작성하는 인텐트 입력 패널, 참조 파일과 히스토리를 보여주는 컨텍스트 서랍, 그리고 심층 코딩이 필요할 때 로컬 IDE로 원클릭 전환할 수 있는 Open IDE 런처를 차례로 안내함.


## Slide 13: Prompt-to-Diff 3단계 실행 생명주기

### 에이전트 실행 3단계 파이프라인
- **Phase 1: Reasoning Plan (추론 및 계획 수립)**
  - 의도 변환(Intent Translation): 사용자의 자연어 프롬프트를 다단계 작업 로드맵으로 세분화하고 의존성을 분석함.
- **Phase 2: Execution Diff (격리된 실행 및 Diff 생성)**
  - 구조화된 실행(Structured Execution): 플랫폼이 격리된 샌드박스 환경에서 파일 수정 사항을 빌드 및 포맷팅함.
- **Phase 3: Human Approval (시각적 검증 및 인간 승인)**
  - 시각적 검증(Visual Inspection): 변경 사항이 로컬 파일에 최종 반영되기 전, 사이드-바이-사이드 코드 diff를 직접 검토하고 승인함.

```mermaid
sequenceDiagram
    autonumber
    actor User as 개발자 (User)
    participant Agent as Antigravity Agent
    participant Sandbox as Server Sandbox
    participant Local as Local Workspace

    User->>Agent: 자연어 프롬프트 입력
    Agent->>Agent: Phase 1: Reasoning Plan 수립 (태스크 로드맵)
    Agent->>Sandbox: Phase 2: 격리 환경에서 코드 수정 및 diff 생성
    Sandbox-->>User: Visual Diff 제시 (사이드-바이-사이드)
    User->>Local: Phase 3: 최종 승인(Human Approval) 후 로컬 파일 적용
```

> **발표자 노트:**
> - 모든 작업은 안전하고 투명한 3단계 생명주기를 거침.
> - 사용자가 프롬프트를 입력하면 에이전트가 단계별 계획을 세우고(Phase 1), 격리 샌드박스에서 코드 diff를 생성하며(Phase 2), 사용자가 직접 눈으로 변경 사항을 확인하고 승인하기 전까지는 로컬 파일이 일체 변경되지 않음(Phase 3).

## Slide 15: 토큰 이코노믹스 및 컨텍스트 최적화 (Token Economics & Context)

### 안티패턴(Anti-Pattern) vs 모범사례(Best Practice) 비교
```text
[ 안티패턴: 전체 코드베이스 덤프 ]
Prompt: "Fix the bug in this 50-file codebase..."
Loaded: src/*, tests/*, config/*, docs/*
Token Usage: 114,250 / 128,000 (92% 용량 소모)
문제점: 긴 응답 지연 시간 발생, 불필요한 노이즈로 인한 에이전트 추론 정확도 저하 및 환각 발생 위험

[ 모범사례: 타겟 컨텍스트 참조 및 커넥터 활용 ]
Prompt: "Fix the auth timeout bug in @auth-service"
Loaded connector: /services/auth/token.go
Token Usage: 3,420 / 128,000 (2.6% 용량 소모)
효과: 초고속 응답 속도 보장, 집중된 컨텍스트로 정밀한 버그 수정 및 높은 추론 품질 달성
```

### 컨텍스트 엔지니어링 3대 핵심 원칙
- **과도한 복사 붙여넣기 지양 (Avoid Over-Pasting)**: 수십 개 파일의 코드를 프롬프트에 직접 붙여넣지 말고, 에이전트의 동적 파일 탐색(Discovery) 기능을 신뢰함.
- **성능 최적화 (Optimize Performance)**: 컨텍스트 윈도우를 간결하게 유지할수록 토큰 비용이 절감되고 응답 지연이 최소화됨.
- **타겟 레퍼런스 활용 (Leverage References)**: `@` 태그나 전용 커넥터를 사용하여 에이전트가 특정 모듈과 인터페이스만 정확히 로드하도록 유도함.

> **발표자 노트:**
> - 첫 사용자들의 가장 흔한 실수는 대규모 코드 전체를 프롬프트에 복사해 넣는 것임.
> - 전체 디렉토리를 쏟아부으면 92%의 컨텍스트를 소모해 느려지고 부정확해지는 반면, 타겟 커넥터를 쓰면 불과 2.6%의 토큰만으로 빠르고 정밀하게 작업을 수행할 수 있음을 강조함.


## Slide 16: 초기 실행 시 빈발 문제 및 트러블슈팅 (Troubleshooting Guide)

### 3대 장애 유형별 해결 절차
| 장애 유형 | 증상 및 원인 | 해결 방안 (Action Item) |
| :--- | :--- | :--- |
| **01. Authentication Timeouts (인증 만료)** | 기업 보안 토큰 만료로 인한 API 호출 거부 | `agy login` 명령어를 실행하여 기업 SSO 재인증 및 세션 갱신 즉시 수행 |
| **02. Permission Warnings (권한 경고)** | 소스코드/저장소 접근 차단 또는 쓰기 권한 부족 | 활성 Google Cloud 프로젝트 바인딩 및 IAM 권한 할당 상태를 재검증함 |
| **03. Session Resets (세션 초기화)** | 복잡한 에이전트 추론 루프가 비정상 정체됨 | 로컬 세션 캐시 상태를 안전하게 초기화(Clean Cache)하여 워크스페이스를 정상 리셋함 |
